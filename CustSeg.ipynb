```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("customer_segments_final.csv")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
df.info()
```

```python
missing_values = df.isnull().sum()

print(missing_values)
```

```python
duplicates = df.duplicated().sum()

print("Duplicate rows:", duplicates)
```

```python
df = df.drop(columns=["Cluster"])

print(df.columns.tolist())
```

```python
df["CustomerID"] = df["CustomerID"].astype(int)

df["Recency"] = df["Recency"].astype(int)

df["Frequency"] = df["Frequency"].astype(int)

df["MonetaryValue"] = df["MonetaryValue"].astype(float)

df.info()
```

```python
print("Customers:", df["CustomerID"].nunique())

print("Minimum Recency:", df["Recency"].min())
print("Maximum Recency:", df["Recency"].max())

print("Minimum Frequency:", df["Frequency"].min())
print("Maximum Frequency:", df["Frequency"].max())

print("Minimum Monetary Value:", df["MonetaryValue"].min())
print("Maximum Monetary Value:", df["MonetaryValue"].max())
```

```python
df.to_csv(
    "customer_rfm_clean.csv",
    index=False
)
print("cleaned customer RFM")
```
